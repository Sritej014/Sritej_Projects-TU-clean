```python
import numpy as np
import matplotlib.pyplot as plt
from pathlib import Path


def load_experiment(path):
    """
    Columns: F11..F33  P11..P33  W
    """
    # ❌ WRONG before:
    # data = np.loadtxt("/Users/.../hyperelasticity/data")

    # ✅ Correct:
    data = np.loadtxt(path)

    F_flat = data[:, 0:9]
    P_flat = data[:, 9:18]
    W = data[:, 18]

    F = F_flat.reshape(-1, 3, 3)
    P = P_flat.reshape(-1, 3, 3)

    return F, P, W


def load_all_data(base_dir):
    # base directory
    base = Path(base_dir)

    calibration_files = {
        "uniaxial":   base / "calibration" / "uniaxial.txt",
        "biaxial":    base / "calibration" / "biaxial.txt",
        "pure_shear": base / "calibration" / "pure_shear.txt",
    }

    test_files = {
        "biax_test":  base / "test" / "biax_test.txt",
        "mixed_test": base / "test" / "mixed_test.txt",
    }

    calib = {}
    for name, f in calibration_files.items():
        F, P, W = load_experiment(f)
        calib[name] = {"F": F, "P": P, "W": W}

    tests = {}
    for name, f in test_files.items():
        F, P, W = load_experiment(f)
        tests[name] = {"F": F, "P": P, "W": W}

    return calib, tests


# -------------------------------------------------
# plotting & classification same as before
# -------------------------------------------------

def plot_load_paths(calib, tests):
    plt.figure(figsize=(6, 6))
    for name, d in calib.items():
        F = d["F"]
        lam1 = F[:, 0, 0]
        lam2 = F[:, 1, 1]
        plt.plot(lam1, lam2, label=f"calib: {name}")

    for name, d in tests.items():
        F = d["F"]
        lam1 = F[:, 0, 0]
        lam2 = F[:, 1, 1]
        plt.plot(lam1, lam2, "--o", markersize=3, label=f"test: {name}")

    plt.xlabel(r"$\lambda_1$")
    plt.ylabel(r"$\lambda_2$")
    plt.grid(True)
    plt.axis("equal")
    plt.legend()

    plt.figure(figsize=(6, 5))
    for name, d in {**calib, **tests}.items():
        F = d["F"]
        P = d["P"]
        gamma = F[:, 0, 1]
        tau = P[:, 0, 1]
        if np.allclose(gamma, 0.0):
            continue
        style = "-" if name in calib else "--"
        plt.plot(gamma, tau, style, label=name)

    plt.xlabel("Shear F12")
    plt.ylabel("P12")
    plt.grid(True)
    plt.legend()
    plt.show()


def classify_interpolation(calib, tests):
    F_calib_flat = np.concatenate(
        [d["F"].reshape(-1, 9) for d in calib.values()], axis=0
    )
    Fmin = F_calib_flat.min(axis=0)
    Fmax = F_calib_flat.max(axis=0)

    for name, d in tests.items():
        F_test_flat = d["F"].reshape(-1, 9)
        inside = np.all((F_test_flat >= Fmin) & (F_test_flat <= Fmax), axis=1)
        frac_inside = inside.mean()
        print(f"{name}: {frac_inside*100:.1f}% inside range")


def main():
    calib, tests = load_all_data(
        "/Users/sritejsritej/Documents/TASK2TutMLSM/hyperelasticity/data"
    )
    plot_load_paths(calib, tests)
    classify_interpolation(calib, tests)


if __name__ == "__main__":
    main()
```

```python
import numpy as np
from pathlib import Path

import jax
import jax.numpy as jnp
from jax import grad, vmap


# ----------------------------------------------------------------------
# 0. Paths
# ----------------------------------------------------------------------
# CHANGE this to your own path if needed:
BASE_DIR = Path(
    "/Users/sritejsritej/Documents/TASK2TutMLSM/hyperelasticity/data"
)

CALIB_DIR = BASE_DIR / "calibration"
TEST_DIR = BASE_DIR / "test"
INV_DIR = BASE_DIR / "invariants"


# ----------------------------------------------------------------------
# 1. Data loading (from txt to numpy, then to jax)
# ----------------------------------------------------------------------
def load_experiment(path: Path):
    """
    Load one text file with columns:
    F11..F33  P11..P33  W
    Returns F, P, W as jax arrays.
    """
    data = np.loadtxt(path)  # (N, 19)

    F_flat = data[:, 0:9]
    P_flat = data[:, 9:18]
    W = data[:, 18]

    F = F_flat.reshape(-1, 3, 3)
    P = P_flat.reshape(-1, 3, 3)

    # convert to jax
    return jnp.array(F), jnp.array(P), jnp.array(W)


def load_all_experiments():
    calibration_files = {
        "uniaxial":   CALIB_DIR / "uniaxial.txt",
        "biaxial":    CALIB_DIR / "biaxial.txt",
        "pure_shear": CALIB_DIR / "pure_shear.txt",
    }

    test_files = {
        "biax_test":  TEST_DIR / "biax_test.txt",
        "mixed_test": TEST_DIR / "mixed_test.txt",
    }

    calib = {}
    for name, path in calibration_files.items():
        F, P, W = load_experiment(path)
        calib[name] = {"F": F, "P": P, "W": W}

    tests = {}
    for name, path in test_files.items():
        F, P, W = load_experiment(path)
        tests[name] = {"F": F, "P": P, "W": W}

    return calib, tests


# ----------------------------------------------------------------------
# 2. Invariants and potential in JAX
# ----------------------------------------------------------------------

# structural tensor G_ti
G_ti = jnp.array(
    [[4.0, 0.0, 0.0],
     [0.0, 0.5, 0.0],
     [0.0, 0.0, 0.5]]
)


def invariants_single(F):
    """
    F : (3,3)
    returns I1, J, I4, I5 (scalars)
    """
    C = F.T @ F                           # right Cauchy-Green
    I1 = jnp.trace(C)

    J = jnp.linalg.det(F)                 # det F

    I4 = jnp.trace(C @ G_ti)              # tr(C G_ti)

    # cof C = det(C) * C^{-T}
    detC = jnp.linalg.det(C)
    CofC = detC * jnp.linalg.inv(C).T
    I5 = jnp.trace(CofC @ G_ti)

    return I1, J, I4, I5


# vectorized version over leading dimension (N,3,3)
invariants = vmap(invariants_single, in_axes=0, out_axes=0)


def W_single(F):
    """
    Hyperelastic potential W(F) = scalar
    """
    I1, J, I4, I5 = invariants_single(F)
    return 8.0 * I1 + 10.0 * J**2 - 56.0 * jnp.log(J) + 0.2 * (I4**2 + I5**2) - 44.0


W_vec = vmap(W_single, in_axes=0, out_axes=0)


# ----------------------------------------------------------------------
# 3. First Piola-Kirchhoff stress P = dW/dF
# ----------------------------------------------------------------------
dW_dF_single = grad(W_single)          # grad wrt F (3x3) -> (3x3)
P_from_W = vmap(dW_dF_single, in_axes=0, out_axes=0)


# ----------------------------------------------------------------------
# 4. Helpers to load invariants from files and compare
#    (you may need to adapt file names in INV_FILES)
# ----------------------------------------------------------------------
# Here I assume a naming convention like:
#   I1_uniaxial.txt, J_uniaxial.txt, I4_uniaxial.txt, I5_uniaxial.txt, etc.
INV_FILES = {
    "uniaxial":   INV_DIR / "I_uniaxial.txt",
    "biaxial":    INV_DIR / "I_biaxial.txt",
    "pure_shear": INV_DIR / "I_pure_shear.txt",
    "biax_test":  INV_DIR / "I_biax_test.txt",
    "mixed_test": INV_DIR / "I_mixed_test.txt",
}



def check_invariants(exp_name, F):
    """
    Compare our JAX invariants with the reference file for this experiment.

    Invariants file format (per Readme):
        I_1   J   I_4   I_5
    -> 4 columns, one row per load step.
    """
    if exp_name not in INV_FILES:
        print(f"[WARN] No invariant file configured for {exp_name}")
        return

    # --- our invariants from JAX ---
    I1, J, I4, I5 = invariants(F)   # each shape: (N,)

    # --- reference invariants from file ---
    ref = jnp.array(np.loadtxt(INV_FILES[exp_name]))  # shape (N, 4)
    ref_I1 = ref[:, 0]
    ref_J  = ref[:, 1]
    ref_I4 = ref[:, 2]
    ref_I5 = ref[:, 3]

    print(f"=== Invariant check for {exp_name} ===")
    print("max|I1 - I1_ref| =", jnp.max(jnp.abs(I1 - ref_I1)))
    print("max|J  - J_ref | =", jnp.max(jnp.abs(J  - ref_J)))
    print("max|I4 - I4_ref| =", jnp.max(jnp.abs(I4 - ref_I4)))
    print("max|I5 - I5_ref| =", jnp.max(jnp.abs(I5 - ref_I5)))
    print()


def check_W_and_P(name, F, P_true, W_true):
    """
    Compare model W and P with data for a given experiment.
    """
    W_model = W_vec(F)
    P_model = P_from_W(F)     # (N,3,3)

    print(f"=== W, P check for {name} ===")
    print("max|W_model - W_true| =", jnp.max(jnp.abs(W_model - W_true)))
    print("max|P_model - P_true| =", jnp.max(jnp.abs(P_model - P_true)))
    print()


# ----------------------------------------------------------------------
# 5. Main routine: run checks on calibration & test data
# ----------------------------------------------------------------------
def main():
    calib, tests = load_all_experiments()

    # calibration
    for name, data in calib.items():
        F, P_true, W_true = data["F"], data["P"], data["W"]
        check_invariants(name, F)
        check_W_and_P(name, F, P_true, W_true)

    # tests – here we can at least check invariants (and W/P if given)
    for name, data in tests.items():
        F, P_true, W_true = data["F"], data["P"], data["W"]
        check_invariants(name, F)
        # if P/W are “truth” values, also:
        check_W_and_P(name, F, P_true, W_true)



if __name__ == "__main__":
    main()
```

```python
import numpy as np
from pathlib import Path

import jax
import jax.numpy as jnp
from jax import grad, jit, random

# ---------- paths ----------
BASE = Path("/Users/sritejsritej/Documents/TASK2TutMLSM/hyperelasticity/data")
CALIB_DIR = BASE / "calibration"
TEST_DIR  = BASE / "test"


def load_experiment(path: Path):
    data = np.loadtxt(path)                 # (N,19)
    F = data[:, :9].reshape(-1, 3, 3)       # (N,3,3)
    P = data[:, 9:18].reshape(-1, 3, 3)     # (N,3,3)
    W = data[:, 18]                         # (N,)
    return jnp.array(F), jnp.array(P), jnp.array(W)


def load_calibration_and_test():
    calib = {
        "uniaxial":   load_experiment(CALIB_DIR / "uniaxial.txt"),
        "biaxial":    load_experiment(CALIB_DIR / "biaxial.txt"),
        "pure_shear": load_experiment(CALIB_DIR / "pure_shear.txt"),
    }
    tests = {
        "biax_test":  load_experiment(TEST_DIR / "biax_test.txt"),
        "mixed_test": load_experiment(TEST_DIR / "mixed_test.txt"),
    }
    return calib, tests
```

```python
#model implementation based functions
# ---------- helpers: C and its 6 independent components ----------

def C_from_F(F):              # F: (N,3,3)
    return jnp.einsum("nij,nkj->nik", F, F)  # F^T F, batched

def C_to_features(C):         # C: (N,3,3) -> (N,6)
    C11 = C[:, 0, 0]
    C22 = C[:, 1, 1]
    C33 = C[:, 2, 2]
    C12 = C[:, 0, 1]
    C13 = C[:, 0, 2]
    C23 = C[:, 1, 2]
    return jnp.stack([C11, C22, C33, C12, C13, C23], axis=1)

def P_to_targets(P):          # P: (N,3,3) -> (N,9)
    return P.reshape(-1, 9)
```

```python
# ffnn degf
# ---------- simple fully-connected network ----------

def init_layer_params(key, in_dim, out_dim):
    k1, k2 = random.split(key)
    W = random.normal(k1, (in_dim, out_dim)) * jnp.sqrt(2.0 / in_dim)
    b = jnp.zeros((out_dim,))
    return W, b

def init_mlp_params(key, in_dim, hidden_dims, out_dim):
    keys = random.split(key, num=len(hidden_dims) + 1)
    params = []
    last_dim = in_dim
    for k, h in zip(keys[:-1], hidden_dims):
        params.append(init_layer_params(k, last_dim, h))
        last_dim = h
    params.append(init_layer_params(keys[-1], last_dim, out_dim))
    return params

def mlp_forward(params, x):
    # x: (N,in_dim)
    for (W, b) in params[:-1]:
        x = jnp.dot(x, W) + b
        x = jax.nn.tanh(x)
    W, b = params[-1]
    x = jnp.dot(x, W) + b
    return x    # (N,out_dim)
```

```python
#objectivity condtion
def model_MS(params, F_batch):
    C = C_from_F(F_batch)
    feats = C_to_features(C)      # (N,6)
    P_pred_flat = mlp_forward(params, feats)  # (N,9)
    return P_pred_flat.reshape(-1, 3, 3)      # (N,3,3)
```

```python
#training on calibration data
def build_calibration_dataset(calib_dict):
    X_list = []
    Y_list = []
    labels  = []   # which load path each sample came from
    for i, (name, (F, P, _W)) in enumerate(calib_dict.items()):
        C = C_from_F(F)
        feats = C_to_features(C)      # (Ni,6)
        targets = P_to_targets(P)     # (Ni,9)
        X_list.append(feats)
        Y_list.append(targets)
        labels.append(jnp.ones((feats.shape[0],), dtype=int) * i)
    X = jnp.concatenate(X_list, axis=0)
    Y = jnp.concatenate(Y_list, axis=0)
    labels = jnp.concatenate(labels, axis=0)
    return X, Y, labels
```

```python
#plain MSE with no wieghing loss
def mse_loss(params, X, Y):
    Y_pred = mlp_forward(params, X)
    return jnp.mean((Y_pred - Y)**2)
```

```python
#simple SGD training
@jit
def update_step(params, X, Y, lr):
    loss_val, grads = jax.value_and_grad(mse_loss)(params, X, Y)
    new_params = [(W - lr * dW, b - lr * db)
                  for (W, b), (dW, db) in zip(params, grads)]
    return new_params, loss_val

def train_naive(calib, num_epochs=5000, lr=5e-4, seed=0):
    X, Y, labels = build_calibration_dataset(calib)
    key = random.PRNGKey(seed)
    params = init_mlp_params(key, in_dim=6, hidden_dims=[32, 32], out_dim=9)

    for epoch in range(num_epochs):
        params, loss_val = update_step(params, X, Y, lr)
        if epoch % 200 == 0:
            print(f"[naive] epoch {epoch}: loss = {loss_val:.4e}")
    return params
```

```python
#Task 2.3 
#weight at each path with punishment for load path based on contribution through weights
def frobenius_norm(P):  # P: (N,3,3)
    return jnp.sqrt(jnp.sum(P**2, axis=(1, 2)))  # (N,)

def compute_path_weights(calib):
    """
    Returns:
      path_names: list in fixed order
      inv_weights: jnp.array of shape (#paths,)
        (inverse L2 weights to use in loss)
    """
    names = []
    inv_ws = []
    for name, (F, P, _W) in calib.items():
        norms = frobenius_norm(P)         # (N,)
        w = jnp.mean(norms)               # scalar
        inv_w = 1.0 / w
        names.append(name)
        inv_ws.append(inv_w)
        print(f"path {name}: w = {float(w):.4e}, 1/w = {float(inv_w):.4e}")
    return names, jnp.array(inv_ws)
```

```python
def weighted_mse_loss(params, X, Y, labels, inv_w):
    """
    params : NN params
    X      : (N, 6)
    Y      : (N, 9)
    labels : (N,) integers in 0..(n_paths-1)
    inv_w  : (n_paths,) inverse weights 1/w_i
    """
    Y_pred = mlp_forward(params, X)
    sq_err = jnp.sum((Y_pred - Y)**2, axis=1)      # (N,)
    w_samples = inv_w[labels]                      # weight of each sample
    weighted_err = w_samples * sq_err             # (N,)

    # normalise by total weight so scale is stable
    return jnp.sum(weighted_err) / jnp.sum(w_samples)
```

```python
#updateing parameters
@jit
def update_step_weighted(params, X, Y, labels, inv_weights, lr):
    loss_val, grads = jax.value_and_grad(weighted_mse_loss)(
        params, X, Y, labels, inv_weights
    )
    new_params = [(W - lr * dW, b - lr * db)
                  for (W, b), (dW, db) in zip(params, grads)]
    return new_params, loss_val
```

```python
#training with loss weighting
def train_weighted(calib, num_epochs=5000, lr=5e-4, seed=1):
    X, Y, labels = build_calibration_dataset(calib)
    path_names, inv_weights = compute_path_weights(calib)

    key = random.PRNGKey(seed)
    params = init_mlp_params(key, in_dim=6, hidden_dims=[32, 32], out_dim=9)

    for epoch in range(num_epochs):
        params, loss_val = update_step_weighted(
            params, X, Y, labels, inv_weights, lr
        )
        if epoch % 200 == 0:
            print(f"[weighted] epoch {epoch}: loss = {loss_val:.4e}")
    return params, path_names, inv_weights
```

```python
#implementaion of model
def main():
    calib, tests = load_calibration_and_test()

    # 1) train naive model
    naive_params = train_naive(calib)

    # 2) train weighted model
    weighted_params, path_names, inv_weights = train_weighted(calib)

    # Example: evaluate on test biaxial
    F_test, P_test, _W_test = tests["biax_test"]
    P_pred_naive   = model_MS(naive_params, F_test)
    P_pred_weighted = model_MS(weighted_params, F_test)
    err_naive = jnp.mean((P_pred_naive - P_test)**2)
    err_weighted = jnp.mean((P_pred_weighted - P_test)**2)
    print("Test biaxial MSE: naive   =", float(err_naive))
    print("Test biaxial MSE: weighted =", float(err_weighted))


if __name__ == "__main__":
    main()
```

```python
#generating plots
import os
import matplotlib.pyplot as plt
import jax.numpy as jnp

# -----------------------------------------------------------
# 1) Helper: compute MSE (Frobenius) for one dataset
# -----------------------------------------------------------
def dataset_mse(params, F, P_true):
    P_pred = model_MS(params, F)              # (N,3,3)
    err = jnp.mean((P_pred - P_true)**2)      # scalar
    return float(err), P_pred


# -----------------------------------------------------------
# 2) Evaluate interpolation (calibration) and test prediction
# -----------------------------------------------------------
def evaluate_models(calib, tests, naive_params, weighted_params):
    print("===== CALIBRATION (interpolation) errors =====")
    for name, (F, P, _W) in calib.items():
        mse_naive, _ = dataset_mse(naive_params, F, P)
        mse_weighted, _ = dataset_mse(weighted_params, F, P)
        print(f"{name:10s} | naive MSE = {mse_naive:.4e} , "
              f"weighted MSE = {mse_weighted:.4e}")

    print("\n===== TEST (unseen load cases) errors =====")
    for name, (F, P, _W) in tests.items():
        mse_naive, _ = dataset_mse(naive_params, F, P)
        mse_weighted, _ = dataset_mse(weighted_params, F, P)
        print(f"{name:10s} | naive MSE = {mse_naive:.4e} , "
              f"weighted MSE = {mse_weighted:.4e}")


# -----------------------------------------------------------
# 3) Plotting utilities
#    For “tension-like” paths: P11 vs lambda1 = F11
#    For “shear-like” paths:   P12 vs gamma = F12
# -----------------------------------------------------------
def choose_x_component(path_name, F):
    """
    Returns x-axis data and which stress component to plot.
    """
    if "pure_shear" in path_name or "mixed" in path_name:
        # shear-dominated: use F12 and P12
        x = F[:, 0, 1]
        comp = (0, 1)
        x_label = r"$\gamma = F_{12}$"
        y_label = r"$P_{12}$"
    else:
        # tension/biaxial: use stretch in 1-dir
        x = F[:, 0, 0]
        comp = (0, 0)
        x_label = r"$\lambda_1 = F_{11}$"
        y_label = r"$P_{11}$"
    return x, comp, x_label, y_label


def plot_load_path(path_name, F, P_true,
                   P_naive, P_weighted,
                   outdir="figs", is_test=False):
    os.makedirs(outdir, exist_ok=True)

    x, comp, x_label, y_label = choose_x_component(path_name, F)
    i, j = comp

    plt.figure(figsize=(6, 4))
    # true data
    plt.plot(x, P_true[:, i, j],
             "o", label="data", markersize=4)
    # naive prediction
    plt.plot(x, P_naive[:, i, j],
             "-", label="naive NN")
    # weighted prediction
    plt.plot(x, P_weighted[:, i, j],
             "--", label="weighted NN")

    kind = "test" if is_test else "calibration"
    plt.title(f"{path_name} ({kind}) : {y_label} vs {x_label}")
    plt.xlabel(x_label)
    plt.ylabel(y_label)
    plt.grid(True)
    plt.legend()
    plt.tight_layout()

    fname = f"{path_name}_{kind}.png"
    plt.savefig(os.path.join(outdir, fname), dpi=200)
    plt.close()


# -----------------------------------------------------------
# 4) Wrapper to generate plots for all paths
# -----------------------------------------------------------
def generate_all_plots(calib, tests, naive_params, weighted_params,
                       outdir="figs"):
    # calibration paths
    for name, (F, P, _W) in calib.items():
        _, P_naive = dataset_mse(naive_params, F, P)
        _, P_weighted = dataset_mse(weighted_params, F, P)
        plot_load_path(name, F, P, P_naive, P_weighted,
                       outdir=outdir, is_test=False)

    # test paths
    for name, (F, P, _W) in tests.items():
        _, P_naive = dataset_mse(naive_params, F, P)
        _, P_weighted = dataset_mse(weighted_params, F, P)
        plot_load_path(name, F, P, P_naive, P_weighted,
                       outdir=outdir, is_test=True)


# -----------------------------------------------------------
# 5) Example main
# -----------------------------------------------------------
def main():
    calib, tests = load_calibration_and_test()

    # train models (or load pre-trained params)
    naive_params = train_naive(calib, num_epochs=5000, lr=5e-04)
    weighted_params, _, _ = train_weighted(calib, num_epochs=5000, lr=5e-4)

    evaluate_models(calib, tests, naive_params, weighted_params)
    generate_all_plots(calib, tests, naive_params, weighted_params,
                       outdir="figs")


if __name__ == "__main__":
    main()
```

```python
import numpy as np
from pathlib import Path

import jax
import jax.numpy as jnp
from jax import random, jit, vmap, grad

import matplotlib.pyplot as plt

BASE_DIR = Path("/Users/sritejsritej/Documents/TASK2TutMLSM/hyperelasticity/data")
CALIB_DIR = BASE_DIR / "calibration"
TEST_DIR  = BASE_DIR / "test"
```

```python
#Load data (F, P, W)
def load_experiment(path: Path):
    data = np.loadtxt(path)                 # (N,19)
    F = data[:, :9].reshape(-1, 3, 3)       # (N,3,3)
    P = data[:, 9:18].reshape(-1, 3, 3)     # (N,3,3)
    W = data[:, 18]                         # (N,)
    return jnp.array(F), jnp.array(P), jnp.array(W)

def load_calib_test():
    calib = {
        "uniaxial":   load_experiment(CALIB_DIR / "uniaxial.txt"),
        "biaxial":    load_experiment(CALIB_DIR / "biaxial.txt"),
        "pure_shear": load_experiment(CALIB_DIR / "pure_shear.txt"),
    }
    tests = {
        "biax_test":  load_experiment(TEST_DIR / "biax_test.txt"),
        "mixed_test": load_experiment(TEST_DIR / "mixed_test.txt"),
    }
    return calib, tests

calib, tests = load_calib_test()
```

```python
#Invariants
G_ti = jnp.array([[4.0, 0.0, 0.0],
                  [0.0, 0.5, 0.0],
                  [0.0, 0.0, 0.5]])

def invariants_single(F):
    C = F.T @ F
    I1 = jnp.trace(C)
    J  = jnp.linalg.det(F)
    I4 = jnp.trace(C @ G_ti)
    detC = jnp.linalg.det(C)
    CofC = detC * jnp.linalg.inv(C).T
    I5 = jnp.trace(CofC @ G_ti)
    return I1, J, I4, I5

def Z_single(F):
    I1, J, I4, I5 = invariants_single(F)
    # <-- key vector used for ICNN
    return jnp.array([I1, J, -J, I4, I5])

Z_batch = vmap(Z_single, in_axes=0, out_axes=0)  # (N,5)
```

```python
#ICNN for energy W
def softplus_pos(x):
    # strictly positive-ish; can use jax.nn.softplus
    return jax.nn.softplus(x)

def init_icnn_params(key, z_dim=5, hidden=[32,32]):
    """
    ICNN params stored as raw weights; positivity enforced via softplus at forward pass.
    """
    keys = random.split(key, 3*len(hidden) + 2)

    params = {}
    # first layer: Wz0 (pos), b0
    params["Wz0_raw"] = random.normal(keys[0], (z_dim, hidden[0])) * 0.1
    params["b0"]      = jnp.zeros((hidden[0],))

    # hidden layers: Wh_raw (pos), Wz_raw (pos), b
    for i in range(1, len(hidden)):
        params[f"Wh{i}_raw"] = random.normal(keys[3*i-2], (hidden[i-1], hidden[i])) * 0.1
        params[f"Wz{i}_raw"] = random.normal(keys[3*i-1], (z_dim, hidden[i])) * 0.1
        params[f"b{i}"]      = jnp.zeros((hidden[i],))

    # output: w_raw (pos), c
    params["w_raw"] = random.normal(keys[-2], (hidden[-1], 1)) * 0.1
    params["c"]     = jnp.zeros((1,))
    return params

def icnn_forward(params, z):
    """
    z: (5,) -> scalar energy W_hat
    """
    # first layer (special restriction: Wz0 >= 0)
    Wz0 = softplus_pos(params["Wz0_raw"])
    h = jax.nn.softplus(z @ Wz0 + params["b0"])   # convex activation

    # other hidden layers
    # (all Wh and Wz are constrained >=0 to preserve convexity in z)
    i = 1
    while f"Wh{i}_raw" in params:
        Wh = softplus_pos(params[f"Wh{i}_raw"])
        Wz = softplus_pos(params[f"Wz{i}_raw"])
        b  = params[f"b{i}"]
        h = jax.nn.softplus(h @ Wh + z @ Wz + b)
        i += 1

    # output is positive combination of convex h => convex
    w = softplus_pos(params["w_raw"])
    W_hat = (h @ w).squeeze() + params["c"].squeeze()
    return W_hat

icnn_forward_batch = vmap(icnn_forward, in_axes=(None, 0), out_axes=0)
```

```python
#physics augumenttion W(Z(F)) ICNN and P = autograd W,F
def W_hat_single(params, F):
    z = Z_single(F)
    return icnn_forward(params, z)

# stress: P = dW/dF
P_hat_single = grad(W_hat_single, argnums=1)

# batched
W_hat_batch = vmap(W_hat_single, in_axes=(None, 0), out_axes=0)
P_hat_batch = vmap(P_hat_single, in_axes=(None, 0), out_axes=0)
```

```python
#stack calibration sets
def stack_dataset(dct):
    F_all, P_all, W_all, label_all = [], [], [], []
    for i,(name,(F,P,W)) in enumerate(dct.items()):
        F_all.append(F); P_all.append(P); W_all.append(W)
        label_all.append(jnp.ones((F.shape[0],), dtype=int)*i)
    return (jnp.concatenate(F_all, axis=0),
            jnp.concatenate(P_all, axis=0),
            jnp.concatenate(W_all, axis=0),
            jnp.concatenate(label_all, axis=0))

Fcal, Pcal, Wcal, Lcal = stack_dataset(calib)
Fte,  Pte,  Wte,  Lte  = stack_dataset(tests)
```

```python
#losses
def frob_norm(P):
    return jnp.sqrt(jnp.sum(P**2, axis=(1,2)))

def compute_inv_weights_from_calib(calib_dict):
    inv_ws = []
    for name,(F,P,W) in calib_dict.items():
        w = jnp.mean(frob_norm(P))
        inv_ws.append(1.0/w)
    inv_ws = jnp.array(inv_ws)
    return inv_ws / jnp.mean(inv_ws)   # normalise for stability

inv_w_paths = compute_inv_weights_from_calib(calib)
```

```python
#loss functions
def mse_energy(params, F, W_true):
    Wp = W_hat_batch(params, F)
    return jnp.mean((Wp - W_true)**2)

def mse_stress(params, F, P_true):
    Pp = P_hat_batch(params, F)
    return jnp.mean((Pp - P_true)**2)

def weighted_mse_stress(params, F, P_true, labels, inv_w):
    Pp = P_hat_batch(params, F)  # (N,3,3)
    se = jnp.sum((Pp - P_true)**2, axis=(1,2))  # (N,)
    w_s = inv_w[labels]
    return jnp.sum(w_s * se) / jnp.sum(w_s)

def loss_energy_only(params, F, P_true, W_true, labels, inv_w, lamW=1.0, lamP=1.0, weighted=True):
    return mse_energy(params, F, W_true)

def loss_stress_only(params, F, P_true, W_true, labels, inv_w, weighted=True):
    if weighted:
        return weighted_mse_stress(params, F, P_true, labels, inv_w)
    else:
        return mse_stress(params, F, P_true)

def loss_both(params, F, P_true, W_true, labels, inv_w, lamW=1.0, lamP=1.0, weighted=True):
    Lw = mse_energy(params, F, W_true)
    Lp = weighted_mse_stress(params, F, P_true, labels, inv_w) if weighted else mse_stress(params, F, P_true)
    return lamW*Lw + lamP*Lp
```

```python
#adam optimiser
def adam_init(params):
    m = jax.tree_util.tree_map(jnp.zeros_like, params)
    v = jax.tree_util.tree_map(jnp.zeros_like, params)
    return m, v

@jit
def adam_step(params, grads, m, v, t, lr=1e-3, b1=0.9, b2=0.999, eps=1e-8):
    m = jax.tree_util.tree_map(lambda m,g: b1*m + (1-b1)*g, m, grads)
    v = jax.tree_util.tree_map(lambda v,g: b2*v + (1-b2)*(g*g), v, grads)
    mhat = jax.tree_util.tree_map(lambda m: m/(1-b1**t), m)
    vhat = jax.tree_util.tree_map(lambda v: v/(1-b2**t), v)
    params = jax.tree_util.tree_map(lambda p,mh,vh: p - lr*mh/(jnp.sqrt(vh)+eps), params, mhat, vhat)
    return params, m, v
```

```python
#training func
def train_icnn(mode="both", num_epochs=5000, lr=1e-3, seed=0, lamW=1.0, lamP=1.0, weighted=True):
    """
    mode: "energy" | "stress" | "both"
    weighted: apply path-weighting to stress loss
    """
    key = random.PRNGKey(seed)
    params = init_icnn_params(key, z_dim=5, hidden=[32,32])
    m, v = adam_init(params)

    if mode == "energy":
        loss_fn = lambda prm: loss_energy_only(prm, Fcal, Pcal, Wcal, Lcal, inv_w_paths, lamW, lamP, weighted)
    elif mode == "stress":
        loss_fn = lambda prm: loss_stress_only(prm, Fcal, Pcal, Wcal, Lcal, inv_w_paths, weighted)
    elif mode == "both":
        loss_fn = lambda prm: loss_both(prm, Fcal, Pcal, Wcal, Lcal, inv_w_paths, lamW, lamP, weighted)
    else:
        raise ValueError("mode must be 'energy', 'stress', or 'both'")

    loss_and_grad = jit(jax.value_and_grad(lambda prm: loss_fn(prm)))

    for epoch in range(1, num_epochs+1):
        loss_val, grads = loss_and_grad(params)
        params, m, v = adam_step(params, grads, m, v, epoch, lr=lr)
        if epoch % 500 == 0:
            print(f"[ICNN {mode}] epoch {epoch}: loss = {float(loss_val):.4e}")
    return params
```

```python
#train em
params_energy = train_icnn(mode="energy", num_epochs=5000, lr=1e-3, seed=0, weighted=True)
params_stress = train_icnn(mode="stress", num_epochs=5000, lr=5e-4, seed=1, weighted=True)
params_both   = train_icnn(mode="both",   num_epochs=5000, lr=5e-4, seed=2, lamW=1.0, lamP=1.0, weighted=True)
```

```python
#evalution helpers
def mse_P(params, F, P_true):
    Pp = P_hat_batch(params, F)
    return jnp.mean((Pp - P_true)**2)

def mse_W(params, F, W_true):
    Wp = W_hat_batch(params, F)
    return jnp.mean((Wp - W_true)**2)

def eval_all(params, title):
    print(f"\n===== {title} =====")
    print("CALIB: MSE_P =", float(mse_P(params, Fcal, Pcal)), " | MSE_W =", float(mse_W(params, Fcal, Wcal)))
    print("TEST : MSE_P =", float(mse_P(params, Fte,  Pte  )), " | MSE_W =", float(mse_W(params, Fte,  Wte  )))

    F_I = jnp.eye(3).reshape(1,3,3)
    W_I = W_hat_batch(params, F_I)[0]
    P_I = P_hat_batch(params, F_I)[0]
    print("W(I) =", float(W_I))
    print("P(I) =\n", np.array(P_I))

eval_all(params_energy, "ICNN trained on ENERGY only")
eval_all(params_stress, "ICNN trained on STRESS only")
eval_all(params_both,   "ICNN trained on BOTH")
```

```python
#Task 4
#load files
import numpy as np
from pathlib import Path

import jax
import jax.numpy as jnp
from jax import random, jit, vmap, grad

import matplotlib.pyplot as plt

BASE_DIR = Path("/Users/sritejsritej/Documents/TASK2TutMLSM/hyperelasticity/data")
CONC_DIR = BASE_DIR / "concentric"
```

```python
#load concentric files 1 in quantity
def load_F_only(path: Path):
    """
    Concentric files typically store only deformation gradients.
    Expected format: 9 columns = F11..F33.
    """
    data = np.loadtxt(path)
    if data.ndim == 1:
        data = data[None, :]
    if data.shape[1] != 9:
        raise ValueError(f"{path.name}: expected 9 columns for F, got {data.shape[1]}")
    F = data.reshape(-1, 3, 3)
    return jnp.array(F)

def list_concentric_paths(conc_dir=CONC_DIR):
    files = sorted(conc_dir.glob("*.txt"))
    if len(files) == 0:
        raise FileNotFoundError(f"No .txt files found in {conc_dir}")
    return files

conc_files = list_concentric_paths()
print("Found load paths:", len(conc_files))
print("Example:", conc_files[0].name)
```

```python
#analytical model of W as per 1.2
G_ti = jnp.array([[4.0, 0.0, 0.0],
                  [0.0, 0.5, 0.0],
                  [0.0, 0.0, 0.5]])

def invariants_single(F):
    C = F.T @ F
    I1 = jnp.trace(C)
    J  = jnp.linalg.det(F)
    I4 = jnp.trace(C @ G_ti)

    detC = jnp.linalg.det(C)
    CofC = detC * jnp.linalg.inv(C).T
    I5 = jnp.trace(CofC @ G_ti)
    return I1, J, I4, I5

def W_analytic_single(F):
    I1, J, I4, I5 = invariants_single(F)
    return 8.0*I1 + 10.0*J**2 - 56.0*jnp.log(J) + 0.2*(I4**2 + I5**2) - 44.0

P_analytic_single = grad(W_analytic_single, argnums=0)

W_analytic_batch = vmap(W_analytic_single, in_axes=0, out_axes=0)
P_analytic_batch = vmap(P_analytic_single, in_axes=0, out_axes=0)
```

```python
#bulid dataset of load path
def build_dataset_from_files(files):
    """
    files: list of paths, each is a load path
    returns concatenated F, P_true, W_true and labels (path index)
    """
    Fs, Ps, Ws, labels = [], [], [], []
    for i, fpath in enumerate(files):
        F = load_F_only(fpath)                  # (Ni,3,3)
        P = P_analytic_batch(F)                 # (Ni,3,3)
        W = W_analytic_batch(F)                 # (Ni,)
        Fs.append(F); Ps.append(P); Ws.append(W)
        labels.append(jnp.ones((F.shape[0],), dtype=int)*i)
    F_all = jnp.concatenate(Fs, axis=0)
    P_all = jnp.concatenate(Ps, axis=0)
    W_all = jnp.concatenate(Ws, axis=0)
    L_all = jnp.concatenate(labels, axis=0)
    return F_all, P_all, W_all, L_all
```

```python
#naive FFNN
def C_from_F(F):
    return jnp.einsum("nij,nkj->nik", F, F)

def C_to_features(C):  # (N,3,3) -> (N,6)
    return jnp.stack([C[:,0,0], C[:,1,1], C[:,2,2], C[:,0,1], C[:,0,2], C[:,1,2]], axis=1)

def P_to_targets(P):   # (N,3,3) -> (N,9)
    return P.reshape(-1, 9)

def init_layer_params(key, in_dim, out_dim):
    W = random.normal(key, (in_dim, out_dim)) * jnp.sqrt(2.0 / in_dim)
    b = jnp.zeros((out_dim,))
    return W, b

def init_mlp_params(key, in_dim=6, hidden=[32,32], out_dim=9):
    keys = random.split(key, len(hidden)+1)
    params=[]
    last=in_dim
    for k,h in zip(keys[:-1], hidden):
        params.append(init_layer_params(k, last, h))
        last=h
    params.append(init_layer_params(keys[-1], last, out_dim))
    return params

def mlp_forward(params, x):
    for W,b in params[:-1]:
        x = jnp.dot(x, W) + b
        x = jax.nn.tanh(x)
    W,b = params[-1]
    return jnp.dot(x, W) + b

def naive_predict_P(params, F):
    C = C_from_F(F)
    X = C_to_features(C)
    Y = mlp_forward(params, X)     # (N,9)
    return Y.reshape(-1,3,3)

def naive_loss(params, F, P_true):
    Pp = naive_predict_P(params, F)
    return jnp.mean((Pp - P_true)**2)

@jit
def naive_step(params, F, P_true, lr):
    loss_val, grads = jax.value_and_grad(naive_loss)(params, F, P_true)
    new_params = [(W - lr*dW, b - lr*db) for (W,b),(dW,db) in zip(params, grads)]
    return new_params, loss_val

def train_naive(F_train, P_train, epochs=3000, lr=1e-3, seed=0, hidden=[32,32]):
    params = init_mlp_params(random.PRNGKey(seed), hidden=hidden)
    for e in range(epochs):
        params, L = naive_step(params, F_train, P_train, lr)
    return params
```

```python
#physics AUG ICNN
def Z_single(F):
    I1, J, I4, I5 = invariants_single(F)
    return jnp.array([I1, J, -J, I4, I5])   # (5,)

Z_batch = vmap(Z_single, in_axes=0, out_axes=0)

def pos(x):
    return jax.nn.softplus(x)

def init_icnn_params(key, z_dim=5, hidden=[32,32]):
    keys = random.split(key, 3*len(hidden) + 2)
    p = {}
    p["Wz0_raw"] = random.normal(keys[0], (z_dim, hidden[0])) * 0.1
    p["b0"] = jnp.zeros((hidden[0],))
    for i in range(1, len(hidden)):
        p[f"Wh{i}_raw"] = random.normal(keys[3*i-2], (hidden[i-1], hidden[i])) * 0.1
        p[f"Wz{i}_raw"] = random.normal(keys[3*i-1], (z_dim, hidden[i])) * 0.1
        p[f"b{i}"] = jnp.zeros((hidden[i],))
    p["w_raw"] = random.normal(keys[-2], (hidden[-1], 1)) * 0.1
    p["c"] = jnp.zeros((1,))
    return p

def icnn_forward(p, z):
    Wz0 = pos(p["Wz0_raw"])
    h = jax.nn.softplus(z @ Wz0 + p["b0"])
    i = 1
    while f"Wh{i}_raw" in p:
        Wh = pos(p[f"Wh{i}_raw"])
        Wz = pos(p[f"Wz{i}_raw"])
        b  = p[f"b{i}"]
        h = jax.nn.softplus(h @ Wh + z @ Wz + b)
        i += 1
    w = pos(p["w_raw"])
    return (h @ w).squeeze() + p["c"].squeeze()

def W_hat_single(p, F):
    z = Z_single(F)
    return icnn_forward(p, z)

P_hat_single = grad(W_hat_single, argnums=1)

W_hat_batch = vmap(W_hat_single, in_axes=(None, 0), out_axes=0)
P_hat_batch = vmap(P_hat_single, in_axes=(None, 0), out_axes=0)

def icnn_loss(params, F, P_true, W_true, lamP=1.0, lamW=1.0):
    Pp = P_hat_batch(params, F)
    Wp = W_hat_batch(params, F)

    LP = jnp.mean((Pp - P_true)**2)
    LW = jnp.mean((Wp - W_true)**2)

    return lamP * LP + lamW * LW

def adam_init(params):
    m = jax.tree_util.tree_map(jnp.zeros_like, params)
    v = jax.tree_util.tree_map(jnp.zeros_like, params)
    return m, v

@jit
def adam_step(params, grads, m, v, t, lr=1e-3, b1=0.9, b2=0.999, eps=1e-8):
    m = jax.tree_util.tree_map(lambda mm,g: b1*mm + (1-b1)*g, m, grads)
    v = jax.tree_util.tree_map(lambda vv,g: b2*vv + (1-b2)*(g*g), v, grads)
    mhat = jax.tree_util.tree_map(lambda mm: mm/(1-b1**t), m)
    vhat = jax.tree_util.tree_map(lambda vv: vv/(1-b2**t), v)
    params = jax.tree_util.tree_map(lambda p,mh,vh: p - lr*mh/(jnp.sqrt(vh)+eps), params, mhat, vhat)
    return params, m, v


@jit
def icnn_step(params, F, P_true, W_true, m, v, t, lr, lamP, lamW):
    L, grads = jax.value_and_grad(icnn_loss)(
        params, F, P_true, W_true, lamP, lamW
    )
    params, m, v = adam_step(params, grads, m, v, t, lr=lr)
    return params, m, v, L


def train_icnn(
    F_train, P_train, W_train,
    epochs=3000, lr=1e-4, seed=0,
    hidden=[32,32], lamP=1.0, lamW=0.0,
    init_params=None
):
    if init_params is None:
        params = init_icnn_params(random.PRNGKey(seed), hidden=hidden)
    else:
        params = init_params

    m, v = adam_init(params)

    for t in range(1, epochs+1):
        params, m, v, L = icnn_step(
            params, F_train, P_train, W_train,
            m, v, t, lr, lamP, lamW
        )

    return params

```

```python
def scale_stress(P):
    scale = jnp.median(jnp.abs(P)) + 1e-12
    return P / scale, scale
```

```python
def Z_single(F):
    I1, J, I4, I5 = invariants_single(F)
    z = jnp.array([I1, J, -J, I4, I5])
    return z / jnp.array([10.0, 1.0, 1.0, 10.0, 10.0])
```

```python


def rel_mse_naive(params, F, P):
    Pp = naive_predict_P(params, F)
    return float(jnp.mean((Pp - P)**2) / (jnp.mean(P**2) + 1e-12))

def rel_mse_icnn(params, F, P):
    Pp = P_hat_batch(params, F)
    return float(jnp.mean((Pp - P)**2) / (jnp.mean(P**2) + 1e-12))


# ------------------------------------------------------------
# GLOBAL SCALING + CACHE
# ------------------------------------------------------------

def compute_global_scale(conc_files):
    vals = []
    for f in conc_files:
        F = load_F_only(f)
        P = P_analytic_batch(F)
        vals.append(jnp.ravel(jnp.abs(P)))
    vals = jnp.concatenate(vals)
    return jnp.median(vals) + 1e-12


def cache_loadpaths(conc_files):
    scale = compute_global_scale(conc_files)
    cache = []
    for f in conc_files:
        F = load_F_only(f)
        P = P_analytic_batch(F)
        W = W_analytic_batch(F)
        cache.append((F, P / scale, W / scale))
    return cache


def build_dataset_from_cache(cache_subset):
    Fs, Ps, Ws = zip(*cache_subset)
    return (
        jnp.concatenate(Fs, axis=0),
        jnp.concatenate(Ps, axis=0),
        jnp.concatenate(Ws, axis=0),
    )


# ------------------------------------------------------------
# ICNN TRAINING (TWO STAGE)
# ------------------------------------------------------------

def train_icnn(
    F_train, P_train, W_train,
    epochs=1500,
    lr=1e-4,
    seed=0,
    hidden=[32, 32],
    lamP=1.0,
    lamW=0.0,
    init_params=None,
):
    if init_params is None:
        params = init_icnn_params(random.PRNGKey(seed), hidden=hidden)
    else:
        params = init_params

    m, v = adam_init(params)

    for t in range(1, epochs + 1):
        params, m, v, _ = icnn_step(
            params,
            F_train,
            P_train,
            W_train,
            m,
            v,
            t,
            lr,
            lamP,
            lamW,
        )

    return params


# ------------------------------------------------------------
# MAIN TASK-4 EXPERIMENT LOOP
# ------------------------------------------------------------

def run_task4_fast(
    cache,
    repeats=5,
    rel_threshold=8e-1,
    naive_hidden=[32, 32],
    icnn_hidden=[32, 32],
    epochs_naive=600,
):

    n_paths = len(cache)
    k_list = list(range(1, n_paths))  # SIMPLE + SAFE

    results = []

    for r in range(repeats):
        print(f"\n========== RUN {r} ==========")

        rng = np.random.default_rng(1000 + r)
        idx = np.arange(n_paths)
        rng.shuffle(idx)
        shuffled = [cache[i] for i in idx]

        curve = []
        k_found = None

        for k in k_list:
            train_cache = shuffled[:k]
            test_cache  = shuffled[k:]

            Ftr, Ptr, Wtr = build_dataset_from_cache(train_cache)
            Fte, Pte, Wte  = build_dataset_from_cache(test_cache)

            # ---- Naive FFNN ----
            naive_params = train_naive(
                Ftr, Ptr,
                epochs=epochs_naive,
                lr=1e-3,
                seed=r,
                hidden=naive_hidden,
            )

            # ---- ICNN stage 1: energy-regularized ----
            icnn_params = train_icnn(
                Ftr, Ptr, Wtr,
                epochs=1500,
                lr=1e-4,
                seed=r,
                hidden=icnn_hidden,
                lamP=1.0,
                lamW=1e-4,
            )

            # ---- ICNN stage 2: stress fine-tuning ----
            icnn_params = train_icnn(
                Ftr, Ptr, Wtr,
                epochs=1500,
                lr=5e-5,
                hidden=icnn_hidden,
                lamP=1.0,
                lamW=0.0,
                init_params=icnn_params,
            )

            naive_err = rel_mse_naive(naive_params, Fte, Pte)
            icnn_err  = rel_mse_icnn(icnn_params,  Fte, Pte)

            curve.append((k, naive_err, icnn_err))

            print(
                f"run {r} | k={k:3d} | "
                f"naive={naive_err:.3e} | icnn={icnn_err:.3e}"
            )

            if naive_err < rel_threshold:
                k_found = k
                break


        results.append({
            "run": r,
            "k_found": k_found,
            "curve": curve,
        })

        print(f"Run {r}: smallest k where BOTH succeed = {k_found}")

    return results


# ------------------------------------------------------------
# RUN TASK 4
# ------------------------------------------------------------

cache = cache_loadpaths(conc_files)

results = run_task4_fast(
    cache,
    repeats=5,
    rel_threshold=8e-1,
    epochs_naive=600,
)
```

```python
import numpy as np
import matplotlib.pyplot as plt

def plot_task4_results(results, threshold=None):
    # Build the union of all k values across runs
    all_ks = sorted({pt[0] for r in results for pt in r["curve"]})
    all_ks = np.array(all_ks, dtype=int)

    n_runs = len(results)
    n_k = len(all_ks)

    naive = np.full((n_runs, n_k), np.nan, dtype=float)
    icnn  = np.full((n_runs, n_k), np.nan, dtype=float)

    # Fill arrays
    for i, r in enumerate(results):
        k_to_vals = {pt[0]: (pt[1], pt[2]) for pt in r["curve"]}
        for j, k in enumerate(all_ks):
            if k in k_to_vals:
                naive[i, j] = k_to_vals[k][0]
                icnn[i, j]  = k_to_vals[k][1]

    # 1) All runs (lines stop where NaNs begin)
    plt.figure(figsize=(7,5))
    for i in range(n_runs):
        plt.plot(all_ks, naive[i], "--", alpha=0.35)
        plt.plot(all_ks, icnn[i],  "-", alpha=0.35)
    if threshold is not None:
        plt.axhline(threshold, linestyle=":")
    plt.yscale("log")
    plt.xlabel("Number of calibration load paths k")
    plt.ylabel("Relative test error (log scale)")
    plt.title("Task 4 — Test error vs k (all runs)")
    plt.grid(True, which="both", ls=":")
    plt.tight_layout()
    plt.show()

    # 2) Mean ± std over runs (ignore NaNs)
    naive_mean = np.nanmean(naive, axis=0)
    naive_std  = np.nanstd(naive, axis=0)
    icnn_mean  = np.nanmean(icnn, axis=0)
    icnn_std   = np.nanstd(icnn, axis=0)

    plt.figure(figsize=(7,5))
    plt.fill_between(all_ks, naive_mean-naive_std, naive_mean+naive_std, alpha=0.25)
    plt.plot(all_ks, naive_mean, "--o", label="Naive (mean)")

    plt.fill_between(all_ks, icnn_mean-icnn_std, icnn_mean+icnn_std, alpha=0.25)
    plt.plot(all_ks, icnn_mean, "-o", label="ICNN (mean)")

    if threshold is not None:
        plt.axhline(threshold, linestyle=":", label="threshold")

    plt.yscale("log")
    plt.xlabel("Number of calibration load paths k")
    plt.ylabel("Relative test error (log scale)")
    plt.title("Task 4 — Mean ± std over runs")
    plt.grid(True, which="both", ls=":")
    plt.legend()
    plt.tight_layout()
    plt.show()

    return all_ks, naive, icnn

# Choose ONE threshold and use it consistently:
THRESH = 8e-1   # e.g. 0.8 (fits your observations)

results = run_task4_fast(
    cache,
    repeats=5,
    rel_threshold=THRESH,
    epochs_naive=600
)

plot_task4_results(results, threshold=THRESH)
```

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from pathlib import Path

# ============================================================
# CONFIG: both H5 stores
# ============================================================
H5_FILES = [
    Path("/Users/sritejsritej/Documents/TASK2TutMLSM/sim-data/BCC/data/deformations.h5"),
    Path("/Users/sritejsritej/Documents/TASK2TutMLSM/sim-data/BCC/data/deformations-1.h5"),
]

F_COLS = ["F11","F12","F13","F21","F22","F23","F31","F32","F33"]
P_COLS = ["P11","P12","P13","P21","P22","P23","P31","P32","P33"]
W_COL  = "StrEn"  # your energy column


# ============================================================
# 1) Enumerate keys from both stores
# ============================================================
def list_keys_in_store(h5_path):
    with pd.HDFStore(h5_path, mode="r") as store:
        return store.keys()

def list_all_keys(h5_files):
    # Return a list of (h5_path, key)
    out = []
    for hp in h5_files:
        keys = list_keys_in_store(hp)
        out.extend([(hp, k) for k in keys])
    return out

all_keys = list_all_keys(H5_FILES)
print("Total load paths across both files:", len(all_keys))
print("Example keys:", all_keys[:10])


# ============================================================
# 2) Load one path -> reshape to matrices (N,3,3)
# ============================================================
def load_path_df(h5_path, key):
    return pd.read_hdf(h5_path, key=key)

def load_path_FPW(h5_path, key):
    df = load_path_df(h5_path, key)

    # ensure required columns exist
    for c in F_COLS + P_COLS + [W_COL]:
        if c not in df.columns:
            raise KeyError(f"Missing column '{c}' in {h5_path}:{key}. Columns={list(df.columns)}")

    F_flat = df[F_COLS].to_numpy(dtype=float)   # (N,9)
    P_flat = df[P_COLS].to_numpy(dtype=float)   # (N,9)
    W      = df[W_COL].to_numpy(dtype=float).reshape(-1)  # (N,)

    F = F_flat.reshape(-1,3,3)
    P = P_flat.reshape(-1,3,3)
    return F, P, W


# ============================================================
# 3) Global scaling factor a (Task 5.1): make P* ~ O(1)
#    We use median(|P|) across ALL paths in BOTH files.
# ============================================================
def compute_global_scale(all_keys, max_paths=None):
    vals = []
    if max_paths is None:
        subset = all_keys
    else:
        subset = all_keys[:max_paths]

    for (hp, k) in subset:
        _, P, _ = load_path_FPW(hp, k)
        vals.append(np.abs(P).ravel())

    vals = np.concatenate(vals)
    a = np.median(vals) + 1e-12
    return float(a)

a = compute_global_scale(all_keys, max_paths=200)  # cap for speed if huge
print("Global scale a =", a)


# ============================================================
# 4) Build cache from keys (scaled)
# ============================================================
def build_cache(all_keys, a):
    cache = []
    for (hp, k) in all_keys:
        F, P, W = load_path_FPW(hp, k)
        cache.append({
            "id": f"{hp.name}:{k}",
            "h5": str(hp),
            "key": k,
            "F": F,
            "P": P / a,
            "W": W / a
        })
    return cache

cache = build_cache(all_keys, a)
print("Cache entries:", len(cache))


# ============================================================
# 5) Task 5.1 Visualization: load paths (F11 vs F22)
# ============================================================
def plot_loadpaths(cache, max_paths=50, title="Load paths (F11 vs F22)"):
    plt.figure(figsize=(6,6))
    for item in cache[:max_paths]:
        F = item["F"]
        lam1 = F[:,0,0]
        lam2 = F[:,1,1]
        plt.plot(lam1, lam2, alpha=0.5)
    plt.xlabel(r"$F_{11}$")
    plt.ylabel(r"$F_{22}$")
    plt.title(title)
    plt.grid(True)
    plt.axis("equal")
    plt.show()

plot_loadpaths(cache, max_paths=50, title="Task 5.1: first 50 load paths (both H5 files)")


# ============================================================
# 6) Concatenate cache subset to arrays
# ============================================================
def concat_cache(cache_subset):
    F = np.concatenate([x["F"] for x in cache_subset], axis=0)
    P = np.concatenate([x["P"] for x in cache_subset], axis=0)
    W = np.concatenate([x["W"] for x in cache_subset], axis=0)
    return F, P, W


# ============================================================
# 7) Relative MSE (scale-invariant metric)
# ============================================================
def rel_mse(P_pred, P_true, eps=1e-12):
    return float(np.mean((P_pred - P_true)**2) / (np.mean(P_true**2) + eps))


# ============================================================
# 8) Task 5 experiment loop:
#    sample k load paths for calibration, rest for test,
#    repeat 5 times, increase k until BOTH succeed.
# ============================================================
# ----------------------------
# Relative MSE
# ----------------------------
def rel_mse(P_pred, P_true, eps=1e-12):
    return float(np.mean((P_pred - P_true)**2) / (np.mean(P_true**2) + eps))


# ----------------------------
# Build k_list (avoid 1..4120 brute force)
# ----------------------------
def make_k_list(n_paths, start=1, growth=1.6, max_k=None):
    if max_k is None:
        max_k = n_paths - 1
    ks = []
    k = start
    while k < max_k:
        ks.append(int(k))
        k = int(np.ceil(k * growth))
        if ks and k == ks[-1]:
            k += 1
    ks = sorted(set([k for k in ks if 1 <= k < n_paths]))
    return ks


# ----------------------------
# Task 5 experiment loop (FIXED: train_icnn gets W_train)
# ----------------------------
def run_task5(
    cache,
    repeats=5,
    threshold=0.2,
    k_list=None,
    epochs_naive=1500,
    # ICNN settings:
    stage1_epochs=1500,
    stage1_lr=1e-4,
    stage1_lamW=1e-4,   # energy regularization weight
    stage2_epochs=1500,
    stage2_lr=5e-5,
    seed0=1000,
):
    """
    Requires in your notebook:
      train_naive(Ftr, Ptr, epochs, lr, seed, hidden?)
      naive_predict_P(params, F)
      train_icnn(Ftr, Ptr, Wtr, epochs, lr, seed, hidden?, lamP, lamW, init_params?)
      P_hat_batch(params, F)
    """
    n = len(cache)
    if k_list is None:
        k_list = make_k_list(n, growth=1.6)   # fast scan

    results = []

    for r in range(repeats):
        print(f"\n========== RUN {r} ==========")
        rng = np.random.default_rng(seed0 + r)
        idx = np.arange(n)
        rng.shuffle(idx)
        shuffled = [cache[i] for i in idx]

        curve = []
        k_found = None

        for k in k_list:
            train_subset = shuffled[:k]
            test_subset  = shuffled[k:]
            if len(test_subset) == 0:
                break

            Ftr = np.concatenate([x["F"] for x in train_subset], axis=0)
            Ptr = np.concatenate([x["P"] for x in train_subset], axis=0)
            Wtr = np.concatenate([x["W"] for x in train_subset], axis=0)

            Fte = np.concatenate([x["F"] for x in test_subset], axis=0)
            Pte = np.concatenate([x["P"] for x in test_subset], axis=0)
            # Wte not needed for error

            # ---- Naive FFNN (stress-only) ----
            naive_params = train_naive(Ftr, Ptr, epochs=epochs_naive, lr=1e-3, seed=r)

            # ---- ICNN Stage 1 (stress + small energy regularization) ----
            icnn_params = train_icnn(
                Ftr, Ptr, Wtr,
                epochs=stage1_epochs,
                lr=stage1_lr,
                seed=r,
                lamP=1.0,
                lamW=stage1_lamW
            )

            # ---- ICNN Stage 2 (stress-only fine-tune) ----
            icnn_params = train_icnn(
                Ftr, Ptr, Wtr,
                epochs=stage2_epochs,
                lr=stage2_lr,
                seed=r,
                lamP=1.0,
                lamW=0.0,
                init_params=icnn_params
            )

            # Predict on test
            Pn = np.array(naive_predict_P(naive_params, Fte))
            Pi = np.array(P_hat_batch(icnn_params, Fte))

            naive_err = rel_mse(Pn, Pte)
            icnn_err  = rel_mse(Pi, Pte)

            curve.append((k, naive_err, icnn_err))
            print(f"run {r} | k={k:4d} | naive={naive_err:.3e} | icnn={icnn_err:.3e}")

            if (naive_err <= threshold) and (icnn_err <= threshold):
                k_found = k
                break

        results.append({"run": r, "k_found": k_found, "curve": curve})
        print(f"Run {r}: smallest k where BOTH succeed = {k_found}")

    return results


# ----------------------------
# Plotting (robust for different curve lengths)
# ----------------------------
def plot_task5_results(results, threshold=None):
    plt.figure(figsize=(7,5))

    for r in results:
        ks = np.array([x[0] for x in r["curve"]], dtype=int)
        naive = np.array([x[1] for x in r["curve"]], dtype=float)
        icnn  = np.array([x[2] for x in r["curve"]], dtype=float)

        plt.plot(ks, naive, "-o", alpha=0.5, label="naive" if r["run"]==0 else None)
        plt.plot(ks, icnn,  "--s", alpha=0.5, label="icnn"  if r["run"]==0 else None)

    if threshold is not None:
        plt.axhline(threshold)

    plt.xlabel("k = number of load paths used for calibration")
    plt.ylabel("Relative MSE on remaining load paths (test)")
    plt.title("Task 5: generalization vs calibration size (both H5 files)")
    plt.grid(True)
    plt.legend()
    plt.show()


# ----------------------------
# RUN
# ----------------------------
THRESH = 0.2
results = run_task5(
    cache,
    repeats=5,
    threshold=THRESH,
    epochs_naive=1500,
    stage1_epochs=1500,
    stage2_epochs=1500,
)
plot_task5_results(results, threshold=THRESH)

```

```python
import numpy as np
import matplotlib.pyplot as plt

# ============================================================
# EDIT HERE: connect to YOUR model prediction functions
# ============================================================
# If your models output scaled P already, keep as-is.
# If your models output unscaled, divide/multiply accordingly.

def model_WI_P(F):   # Section 5.2 invariant-based model
    return np.array(predict_WI_P(params_WI, F))

def model_WF_P(F):   # Section 5.3 F/cofF/detF model
    return np.array(predict_WF_P(params_WF, F))

# Optional (if you trained energy models)
# def model_WI_W(F): return np.array(predict_WI_W(params_WI, F))
# def model_WF_W(F): return np.array(predict_WF_W(params_WF, F))


# ============================================================
# Helper: categorize load paths from key names (for plotting)
# ============================================================
def key_type(item):
    k = item["key"].lower()
    if "uniax" in k: return "uniaxial"
    if "shear_simple" in k: return "shear_simple"
    if "shear_combined" in k: return "shear_combined"
    if "biax" in k: return "biaxial"
    if "biplanar" in k: return "biplanar"
    if "planar" in k: return "planar"
    if "volumetric" in k: return "volumetric"
    return "other"

def group_cache_by_type(cache, max_per_type=200):
    groups = {}
    for it in cache:
        t = key_type(it)
        if t not in groups:
            groups[t] = []
        if len(groups[t]) < max_per_type:
            groups[t].append(it)
    return groups


# ============================================================
# Helper: sample points from a path for scatter plots
# ============================================================
def sample_path_points(F, P=None, W=None, n=200):
    N = F.shape[0]
    if N <= n:
        idx = np.arange(N)
    else:
        idx = np.linspace(0, N-1, n).astype(int)
    if P is None and W is None:
        return F[idx]
    out = [F[idx]]
    if P is not None: out.append(P[idx])
    if W is not None: out.append(W[idx])
    return out


# ============================================================
# 5.1 PLOTS: Data preparation + scaling + load path visualization
# ============================================================

def plot_51_loadpaths_F11_F22(cache, max_paths=80):
    plt.figure(figsize=(6,6))
    for it in cache[:max_paths]:
        F = it["F"]
        plt.plot(F[:,0,0], F[:,1,1], alpha=0.5)
    plt.xlabel(r"$F_{11}$")
    plt.ylabel(r"$F_{22}$")
    plt.title("Task 5.1: Load paths in stretch space (F11 vs F22)")
    plt.grid(True); plt.axis("equal")
    plt.show()

def plot_51_hist_P_before_after_scale(cache, a, n_paths=80, n_samples_per_path=200):
    # gather raw P and scaled P for some paths
    raw_vals = []
    scaled_vals = []
    for it in cache[:n_paths]:
        # cache stores scaled P already (P/a). Recover raw:
        P_scaled = it["P"]
        P_raw = P_scaled * a
        Pr = sample_path_points(it["F"], P=P_raw, n=n_samples_per_path)[1]
        Ps = sample_path_points(it["F"], P=P_scaled, n=n_samples_per_path)[1]
        raw_vals.append(np.abs(Pr).ravel())
        scaled_vals.append(np.abs(Ps).ravel())
    raw_vals = np.concatenate(raw_vals)
    scaled_vals = np.concatenate(scaled_vals)

    plt.figure(figsize=(7,4))
    plt.hist(raw_vals, bins=80, alpha=0.6, label="|P| raw")
    plt.hist(scaled_vals, bins=80, alpha=0.6, label="|P| scaled (P/a)")
    plt.xlabel("absolute stress component magnitude")
    plt.ylabel("count")
    plt.title("Task 5.1: Output scaling check (raw vs scaled)")
    plt.grid(True)
    plt.legend()
    plt.show()


# ============================================================
# 5.2 PLOTS: invariant-based model performance
#   - parity plot (P_true vs P_pred)
#   - component curves along specific load types (uniaxial + shear)
# ============================================================

def plot_parity_P(modelP, cache_subset, a=1.0, title="Parity plot", max_points=20000):
    # cache_subset P is scaled; compare scaled-to-scaled
    pts_true = []
    pts_pred = []

    for it in cache_subset:
        F, P, _ = it["F"], it["P"], it["W"]
        F_s, P_s = sample_path_points(F, P=P, n=300)
        P_hat = modelP(F_s)
        pts_true.append(P_s.reshape(-1))
        pts_pred.append(P_hat.reshape(-1))

    y = np.concatenate(pts_true)
    x = np.concatenate(pts_pred)

    # limit points for speed/clarity
    if x.size > max_points:
        idx = np.random.default_rng(0).choice(x.size, size=max_points, replace=False)
        x = x[idx]; y = y[idx]

    plt.figure(figsize=(5.5,5.5))
    plt.scatter(y, x, s=3, alpha=0.25)
    lo = min(y.min(), x.min()); hi = max(y.max(), x.max())
    plt.plot([lo,hi],[lo,hi])
    plt.xlabel("P true (scaled)")
    plt.ylabel("P predicted (scaled)")
    plt.title(title)
    plt.grid(True)
    plt.axis("equal")
    plt.show()

def plot_component_along_path(modelP, it, comp=("11"), title=""):
    # comp "11" -> P11, "12"->P12 etc
    i = int(comp[0]) - 1
    j = int(comp[1]) - 1

    F = it["F"]
    P = it["P"]
    # x-axis: use F11 or shear gamma depending on comp
    if (i,j) == (0,1) or (i,j) == (1,0):
        x = F[:,0,1]  # gamma
        xlabel = r"$\gamma = F_{12}$"
    else:
        x = F[:,0,0]  # stretch
        xlabel = r"$F_{11}$"

    P_hat = modelP(F)
    plt.figure(figsize=(7,4))
    plt.plot(x, P[:,i,j], label=f"true P{comp}")
    plt.plot(x, P_hat[:,i,j], "--", label=f"pred P{comp}")
    plt.xlabel(xlabel)
    plt.ylabel(f"P{comp} (scaled)")
    plt.title(title)
    plt.grid(True)
    plt.legend()
    plt.show()


# ============================================================
# 5.3 PLOTS: objectivity check across multiple observers (rotations)
#   For objective model: P(QF) ≈ Q P(F)
#   Plot violation norm vs rotation index (or histogram)
# ============================================================

def random_rotations(n=16, seed=0):
    rng = np.random.default_rng(seed)
    Qs = []
    for _ in range(n):
        # random unit quaternion -> rotation
        u1,u2,u3 = rng.random(3)
        q1 = np.sqrt(1-u1)*np.sin(2*np.pi*u2)
        q2 = np.sqrt(1-u1)*np.cos(2*np.pi*u2)
        q3 = np.sqrt(u1)*np.sin(2*np.pi*u3)
        q4 = np.sqrt(u1)*np.cos(2*np.pi*u3)
        # rotation matrix
        x,y,z,w = q1,q2,q3,q4
        Q = np.array([
            [1-2*(y*y+z*z), 2*(x*y-z*w),   2*(x*z+y*w)],
            [2*(x*y+z*w),   1-2*(x*x+z*z), 2*(y*z-x*w)],
            [2*(x*z-y*w),   2*(y*z+x*w),   1-2*(x*x+y*y)]
        ])
        Qs.append(Q)
    return np.stack(Qs, axis=0)  # (n,3,3)

def objectivity_violation(modelP, F, Qs):
    """
    returns violations for each Q:
      v(Q) = || P(QF) - Q P(F) ||_F / (||P(QF)||_F + eps)
    """
    P0 = modelP(F)  # (N,3,3)
    viols = []

    for Q in Qs:
        Fq = (Q @ F.transpose(0,1,2)).transpose(0,1,2)  # QF for each sample
        Pq = modelP(Fq)
        QP0 = (Q @ P0.transpose(0,1,2)).transpose(0,1,2)

        num = np.linalg.norm((Pq - QP0).reshape(F.shape[0], -1), axis=1)
        den = np.linalg.norm(Pq.reshape(F.shape[0], -1), axis=1) + 1e-12
        viols.append(np.mean(num/den))
    return np.array(viols)

def plot_objectivity(modelP, cache, n_paths=5, n_rot=16, seed=0, title="Objectivity check"):
    Qs = random_rotations(n_rot, seed=seed)
    # take some representative paths
    picks = cache[::max(1, len(cache)//n_paths)][:n_paths]

    plt.figure(figsize=(7,4))
    for idx, it in enumerate(picks):
        F = it["F"]
        # sample for speed
        F_s = sample_path_points(F, n=200)
        v = objectivity_violation(modelP, F_s, Qs)
        plt.plot(np.arange(len(v)), v, "-o", alpha=0.8, label=f"path {idx}")
    plt.xlabel("rotation index")
    plt.ylabel("mean relative violation")
    plt.title(title)
    plt.grid(True)
    plt.legend()
    plt.show()


# ============================================================
# 5.4 PLOTS: data augmentation effect
#   run with different n_rot in augmentation set and plot violation vs n_rot
# ============================================================

def plot_augmented_objectivity_curve(model_builder_train_fn,  # you provide
                                     cache_train,
                                     cache_test,
                                     n_rot_list=(8,16,32),
                                     obj_check_rot=32):
    """
    model_builder_train_fn(n_rot) must:
      - build augmentation with n_rot rotations
      - train WF model
      - return trained params (or a callable modelP)
    """
    Qcheck = random_rotations(obj_check_rot, seed=123)
    means = []

    for n_rot in n_rot_list:
        print(f"Training with n_rot={n_rot} ...")
        modelP = model_builder_train_fn(n_rot)  # must return callable: modelP(F)->P_hat

        # evaluate objectivity violation on test set (few paths)
        picks = cache_test[::max(1, len(cache_test)//8)][:8]
        v_all = []
        for it in picks:
            F = sample_path_points(it["F"], n=200)
            v = objectivity_violation(modelP, F, Qcheck)
            v_all.append(v.mean())
        means.append(np.mean(v_all))

    plt.figure(figsize=(6,4))
    plt.plot(list(n_rot_list), means, "-o")
    plt.xlabel("# rotations used for augmentation (|G_obj|)")
    plt.ylabel("mean objectivity violation on test")
    plt.title("Task 5.4: augmentation improves objectivity (should decrease)")
    plt.grid(True)
    plt.show()


# ============================================================
# RUN PLOTS
# ============================================================

# ---- 5.1
plot_51_loadpaths_F11_F22(cache, max_paths=120)
plot_51_hist_P_before_after_scale(cache, a=a, n_paths=80)

# ---- 5.2 (needs params_WI + predict_WI_P)
groups = group_cache_by_type(cache, max_per_type=50)
if "uniaxial" in groups:
    plot_parity_P(model_WI_P, groups["uniaxial"], title="Task 5.2: WI parity (uniaxial subset)")
if "shear_simple" in groups:
    plot_parity_P(model_WI_P, groups["shear_simple"], title="Task 5.2: WI parity (shear_simple subset)")

# example component curve on ONE path
if "uniaxial" in groups and len(groups["uniaxial"]) > 0:
    plot_component_along_path(model_WI_P, groups["uniaxial"][0], comp="11",
                              title="Task 5.2: WI on one uniaxial path (P11 vs F11)")
if "shear_simple" in groups and len(groups["shear_simple"]) > 0:
    plot_component_along_path(model_WI_P, groups["shear_simple"][0], comp="12",
                              title="Task 5.2: WI on one shear path (P12 vs F12)")

# ---- 5.3 (needs params_WF + predict_WF_P)
plot_objectivity(model_WF_P, cache, n_paths=6, n_rot=16, seed=0,
                 title="Task 5.3: Objectivity violation for WF model (multi-observer)")

# ---- 5.4: augmentation curve
# You MUST provide the function model_builder_train_fn(n_rot).
# Example placeholder (will error until you implement it):
# plot_augmented_objectivity_curve(model_builder_train_fn, cache_train, cache_test,
#                                  n_rot_list=(8,16,32), obj_check_rot=32)
```

```python
import numpy as np
import jax
import jax.numpy as jnp
from jax import random, vmap, jit, grad
import matplotlib.pyplot as plt

# ============================================================
# 0) Utilities: cof(F), det(F), features
# ============================================================

def cof3x3(F):
    # cofactor of 3x3: cof(F) = det(F) * inv(F)^T
    return jnp.linalg.det(F) * jnp.linalg.inv(F).T

def features_F_cof_det(F):
    # F: (...,3,3) -> (...,19)
    cofF = vmap(cof3x3)(F) if F.ndim == 3 else cof3x3(F)
    detF = jnp.linalg.det(F) if F.ndim == 2 else jnp.linalg.det(F)[:, None]
    if F.ndim == 2:
        x = jnp.concatenate([F.reshape(-1), cofF.reshape(-1), detF.reshape(-1)])
    else:
        x = jnp.concatenate([F.reshape(F.shape[0], 9),
                             cofF.reshape(F.shape[0], 9),
                             detF.reshape(F.shape[0], 1)], axis=1)
    return x

# ============================================================
# 1) Random rotations SO(3) for observers + augmentation
# ============================================================

def random_rotations(n, seed=0):
    rng = np.random.default_rng(seed)
    Qs = []
    for _ in range(n):
        u1, u2, u3 = rng.random(3)
        q1 = np.sqrt(1-u1)*np.sin(2*np.pi*u2)
        q2 = np.sqrt(1-u1)*np.cos(2*np.pi*u2)
        q3 = np.sqrt(u1)*np.sin(2*np.pi*u3)
        q4 = np.sqrt(u1)*np.cos(2*np.pi*u3)
        x,y,z,w = q1,q2,q3,q4
        Q = np.array([
            [1-2*(y*y+z*z), 2*(x*y-z*w),   2*(x*z+y*w)],
            [2*(x*y+z*w),   1-2*(x*x+z*z), 2*(y*z-x*w)],
            [2*(x*z-y*w),   2*(y*z+x*w),   1-2*(x*x+y*y)]
        ], dtype=float)
        Qs.append(Q)
    return jnp.array(np.stack(Qs, axis=0))  # (n,3,3)

def apply_Q_to_batch(Q, A):
    # Q: (3,3), A: (N,3,3) -> (N,3,3) with left multiplication
    return jnp.einsum("ij,njk->nik", Q, A)

# ============================================================
# 2) WF model: ICNN (convex in x) -> scalar W(F)
#    Standard ICNN: z_{l+1} = softplus( W_x x + W_z z_l + b )
#    with W_z >= 0 for convexity in x
# ============================================================

def softplus(x):
    return jnp.log1p(jnp.exp(-jnp.abs(x))) + jnp.maximum(x, 0)

def init_icnn_params(key, in_dim=19, hidden=(64,64)):
    keys = random.split(key, 10)
    params = []
    prev = in_dim
    # first layer: affine in x only
    h0 = hidden[0]
    Wx0 = 0.1 * random.normal(keys[0], (h0, prev))
    b0  = jnp.zeros((h0,))
    params.append({"Wx": Wx0, "Wz": None, "b": b0})

    # hidden layers: affine in x and z, with Wz >= 0
    for li, h in enumerate(hidden[1:], start=1):
        Wx = 0.1 * random.normal(keys[li], (h, in_dim))
        Wz_raw = 0.1 * random.normal(keys[li+1], (h, hidden[li-1]))
        b  = jnp.zeros((h,))
        params.append({"Wx": Wx, "Wz_raw": Wz_raw, "b": b})

    # output layer: scalar W, affine in x and z_last, with Wz >= 0
    Wx_out = 0.1 * random.normal(keys[-2], (1, in_dim))
    Wz_out_raw = 0.1 * random.normal(keys[-1], (1, hidden[-1]))
    b_out = jnp.zeros((1,))
    params.append({"Wx": Wx_out, "Wz_raw": Wz_out_raw, "b": b_out})
    return params

def icnn_forward(params, x):
    # x: (19,)
    # layer 0
    z = softplus(params[0]["Wx"] @ x + params[0]["b"])
    # hidden
    for layer in params[1:-1]:
        Wz = softplus(layer["Wz_raw"])  # enforce >= 0
        z = softplus(layer["Wx"] @ x + Wz @ z + layer["b"])
    # output
    out_layer = params[-1]
    Wz = softplus(out_layer["Wz_raw"])
    y = (out_layer["Wx"] @ x + Wz @ z + out_layer["b"])[0]
    return y  # scalar

def W_WF(params, F):
    # F: (3,3)
    x = features_F_cof_det(F)  # (19,)
    return icnn_forward(params, x)

# Stress via grad wrt F
dW_dF = jit(grad(W_WF, argnums=1))
P_WF_single = jit(lambda params, F: dW_dF(params, F))

P_WF_batch = jit(vmap(P_WF_single, in_axes=(None, 0)))  # (N,3,3)
W_WF_batch = jit(vmap(W_WF, in_axes=(None, 0)))         # (N,)

# ============================================================
# 3) Optimizer (Adam) + training loop
# ============================================================

def adam_init(params):
    m = jax.tree_util.tree_map(jnp.zeros_like, params)
    v = jax.tree_util.tree_map(jnp.zeros_like, params)
    return m, v

def adam_update(params, grads, m, v, t, lr, b1=0.9, b2=0.999, eps=1e-8):
    m = jax.tree_util.tree_map(lambda m,g: b1*m + (1-b1)*g, m, grads)
    v = jax.tree_util.tree_map(lambda v,g: b2*v + (1-b2)*(g*g), v, grads)
    mhat = jax.tree_util.tree_map(lambda m: m/(1-b1**t), m)
    vhat = jax.tree_util.tree_map(lambda v: v/(1-b2**t), v)
    params = jax.tree_util.tree_map(lambda p,mh,vh: p - lr*mh/(jnp.sqrt(vh)+eps),
                                    params, mhat, vhat)
    return params, m, v

def loss_WF(params, F, P_true, W_true=None, lamP=1.0, lamW=0.0):
    P_pred = P_WF_batch(params, F)
    LP = jnp.mean((P_pred - P_true)**2)
    if (W_true is None) or (lamW == 0.0):
        return lamP * LP
    W_pred = W_WF_batch(params, F)
    LW = jnp.mean((W_pred - W_true)**2)
    return lamP * LP + lamW * LW

loss_grad_WF = jit(jax.value_and_grad(loss_WF))

def train_WF(F, P, W, epochs=2000, lr=1e-4, seed=0, lamP=1.0, lamW=1e-4):
    params = init_icnn_params(random.PRNGKey(seed), in_dim=19, hidden=(64,64))
    m, v = adam_init(params)

    for t in range(1, epochs+1):
        L, grads = loss_grad_WF(params, F, P, W, lamP, lamW)
        params, m, v = adam_update(params, grads, m, v, t, lr)
        if (t % 500) == 0:
            print(f"[WF] epoch {t}: loss={float(L):.3e}")
    return params

# ============================================================
# 4) 5.4 Data augmentation: D -> D_tilde using Q in G_obj
# ============================================================

def augment_dataset_objectivity(F, P, W, Qs):
    # D_tilde = union over Q: (QF, W, QP)
    F_list = [F]
    P_list = [P]
    W_list = [W]
    for Q in Qs:
        Fq = apply_Q_to_batch(Q, F)
        Pq = apply_Q_to_batch(Q, P)
        F_list.append(Fq)
        P_list.append(Pq)
        W_list.append(W)  # energy is objective: W(QF)=W(F)
    return (jnp.concatenate(F_list, axis=0),
            jnp.concatenate(P_list, axis=0),
            jnp.concatenate(W_list, axis=0))

# ============================================================
# 5) Objectivity violation metric (Sec 5.3 / 5.4 evaluation)
# ============================================================

def objectivity_violation_WF(params, F, Qs):
    # v(Q) = mean ||P(QF) - Q P(F)|| / (||P(QF)|| + eps)
    P0 = P_WF_batch(params, F)
    vals = []
    for Q in Qs:
        Fq = apply_Q_to_batch(Q, F)
        Pq = P_WF_batch(params, Fq)
        QP0 = apply_Q_to_batch(Q, P0)
        num = jnp.linalg.norm((Pq - QP0).reshape(F.shape[0], -1), axis=1)
        den = jnp.linalg.norm(Pq.reshape(F.shape[0], -1), axis=1) + 1e-12
        vals.append(jnp.mean(num/den))
    return jnp.array(vals)

# ============================================================
# 6) Train WF with augmentation sizes and plot results (Task 5.4)
# ============================================================

def split_cache(cache, frac_train=0.7, seed=0):
    rng = np.random.default_rng(seed)
    idx = np.arange(len(cache))
    rng.shuffle(idx)
    m = int(frac_train * len(cache))
    train_idx, test_idx = idx[:m], idx[m:]
    train = [cache[i] for i in train_idx]
    test  = [cache[i] for i in test_idx]
    return train, test

def concat_cache(cache_subset, max_paths=None):
    if max_paths is not None:
        cache_subset = cache_subset[:max_paths]
    F = jnp.concatenate([it["F"] for it in cache_subset], axis=0)
    P = jnp.concatenate([it["P"] for it in cache_subset], axis=0)
    W = jnp.concatenate([it["W"] for it in cache_subset], axis=0)
    return F, P, W

def run_task54(cache, n_rot_list=(0,8,16,32), epochs=2000, lr=1e-4,
               lamW=1e-4, seed=0, max_train_paths=200, max_test_paths=40):
    train_cache, test_cache = split_cache(cache, frac_train=0.7, seed=seed)
    Ftr, Ptr, Wtr = concat_cache(train_cache, max_paths=max_train_paths)
    Fte, Pte, Wte = concat_cache(test_cache,  max_paths=max_test_paths)

    Qcheck = random_rotations(32, seed=123)  # fixed for comparison

    out = []
    for n_rot in n_rot_list:
        print(f"\n=== Train WF with augmentation |G_obj|={n_rot} ===")
        if n_rot == 0:
            Faug, Paug, Waug = Ftr, Ptr, Wtr
        else:
            Qs = random_rotations(n_rot, seed=999 + n_rot)
            Faug, Paug, Waug = augment_dataset_objectivity(Ftr, Ptr, Wtr, Qs)

        params = train_WF(Faug, Paug, Waug, epochs=epochs, lr=lr, seed=seed, lamP=1.0, lamW=lamW)

        # test error (relative MSE on stress)
        Ppred = np.array(P_WF_batch(params, Fte))
        rel = float(np.mean((Ppred - np.array(Pte))**2) / (np.mean(np.array(Pte)**2) + 1e-12))

        # objectivity violation on test (few samples)
        F_small = Fte[::max(1, Fte.shape[0]//300)]  # ~300 points
        v = np.array(objectivity_violation_WF(params, F_small, Qcheck))
        vmean = float(v.mean())

        out.append({"n_rot": n_rot, "params": params, "rel_mse_test": rel, "viol": v, "viol_mean": vmean})
        print(f"Test relMSE(P) = {rel:.3e} | mean objectivity violation = {vmean:.3e}")

    return out

def plot_task54(results):
    n_rot = [r["n_rot"] for r in results]
    viol_mean = [r["viol_mean"] for r in results]
    relmse = [r["rel_mse_test"] for r in results]

    plt.figure(figsize=(6,4))
    plt.plot(n_rot, viol_mean, "-o")
    plt.xlabel("|G_obj| (# rotations used in augmentation)")
    plt.ylabel("mean objectivity violation on test")
    plt.title("Task 5.4: Objectivity improves with augmentation")
    plt.grid(True)
    plt.show()

    plt.figure(figsize=(6,4))
    plt.plot(n_rot, relmse, "-o")
    plt.xlabel("|G_obj| (# rotations used in augmentation)")
    plt.ylabel("test relative MSE(P)")
    plt.title("Task 5.4: Test error vs augmentation size")
    plt.grid(True)
    plt.show()

    # optional: show violation curves for each augmentation size
    plt.figure(figsize=(7,4))
    for r in results:
        plt.plot(np.arange(len(r["viol"])), r["viol"], label=f"|G|={r['n_rot']}")
    plt.xlabel("observer rotation index")
    plt.ylabel("mean violation for that observer")
    plt.title("Task 5.4: Multi-observer objectivity violation curves")
    plt.grid(True)
    plt.legend()
    plt.show()

# ============================================================
# RUN TASK 5.4
# ============================================================
# NOTE: this can be heavy. Start small, then increase epochs/max_paths.
task54_results = run_task54(
    cache,
    n_rot_list=(0, 8, 16, 32),
    epochs=1500,
    lr=1e-4,
    lamW=1e-4,
    seed=0,
    max_train_paths=200,   # increase later (e.g. 500, 1000)
    max_test_paths=40      # increase later
)

plot_task54(task54_results)
```

```python
import numpy as np
import matplotlib.pyplot as plt

# ============================================================
# Helper: safe subsampling so plots never hang
# ============================================================
def subsample_idx(n, max_n=2000, seed=0):
    if n <= max_n:
        return np.arange(n)
    rng = np.random.default_rng(seed)
    return rng.choice(n, size=max_n, replace=False)

def flatten_tensor(T):
    # (N,3,3) -> (N,9)
    return T.reshape(T.shape[0], 9)

# ============================================================
# 5.1 PLOTS: dataset + scaling sanity + stress magnitude
# ============================================================
def plot_task51(cache, max_paths=60, max_pts=1500):
    # 5.1a: load paths in stretch plane
    plt.figure(figsize=(6,6))
    for item in cache[:max_paths]:
        F = item["F"]
        plt.plot(F[:,0,0], F[:,1,1], alpha=0.5)
    plt.xlabel("F11")
    plt.ylabel("F22")
    plt.title("Task 5.1a: Load paths (F11 vs F22)")
    plt.grid(True)
    plt.axis("equal")
    plt.show()

    # 5.1b: stress magnitude distribution after scaling
    # Collect a sample of stress values
    Ps = []
    for item in cache[:max_paths]:
        P = item["P"]  # already scaled by /a
        Ps.append(np.abs(P).ravel())
    Ps = np.concatenate(Ps)
    idx = subsample_idx(Ps.size, max_n=max_pts, seed=0)

    plt.figure(figsize=(6,4))
    plt.hist(Ps[idx], bins=60)
    plt.xlabel("|P*| values (scaled)")
    plt.ylabel("count")
    plt.title("Task 5.1b: Check scaling — P* should be ~O(1)")
    plt.grid(True)
    plt.show()

    # 5.1c: energy distribution after scaling
    Ws = []
    for item in cache[:max_paths]:
        Ws.append(item["W"].ravel())
    Ws = np.concatenate(Ws)
    idx = subsample_idx(Ws.size, max_n=max_pts, seed=1)

    plt.figure(figsize=(6,4))
    plt.hist(Ws[idx], bins=60)
    plt.xlabel("W* values (scaled)")
    plt.ylabel("count")
    plt.title("Task 5.1c: Scaled energy distribution (W*)")
    plt.grid(True)
    plt.show()


# ============================================================
# 5.2 PLOTS: invariant-based model evaluation
# - Train on a subset of load paths, test on the rest
# - Show parity plots for key components of P
# ============================================================
def plot_task52_invariant_model(cache, k_train=50, epochs=2000, seed=0, max_test_pts=3000):
    rng = np.random.default_rng(seed)
    idx = np.arange(len(cache))
    rng.shuffle(idx)
    train_ids = idx[:k_train]
    test_ids  = idx[k_train:]

    train_subset = [cache[i] for i in train_ids]
    test_subset  = [cache[i] for i in test_ids]

    Ftr = np.concatenate([x["F"] for x in train_subset], axis=0)
    Ptr = np.concatenate([x["P"] for x in train_subset], axis=0)
    Wtr = np.concatenate([x["W"] for x in train_subset], axis=0)

    Fte = np.concatenate([x["F"] for x in test_subset], axis=0)
    Pte = np.concatenate([x["P"] for x in test_subset], axis=0)
    Wte = np.concatenate([x["W"] for x in test_subset], axis=0)

    # ---- train invariant ICNN (your existing one) ----
    icnn_params = train_icnn(
        Ftr, Ptr, Wtr,
        epochs=epochs,
        lr=1e-4,
        seed=seed,
        lamP=1.0,
        lamW=1e-4
    )

    # predict on test
    Ppred = np.array(P_hat_batch(icnn_params, Fte))

    # subsample points for plotting
    n = Pte.shape[0]
    take = subsample_idx(n, max_n=max_test_pts, seed=seed)

    # parity plots for a few components
    comps = {
        "P11": (0,0),
        "P22": (1,1),
        "P12": (0,1),
        "P33": (2,2),
    }

    plt.figure(figsize=(7,6))
    for name,(i,j) in comps.items():
        plt.scatter(Pte[take,i,j], Ppred[take,i,j], s=6, alpha=0.35, label=name)

    mn = min(Pte[take].min(), Ppred[take].min())
    mx = max(Pte[take].max(), Ppred[take].max())
    plt.plot([mn,mx],[mn,mx])
    plt.xlabel("True P*")
    plt.ylabel("Predicted P*")
    plt.title(f"Task 5.2: Invariant ICNN parity (train k={k_train} load paths)")
    plt.grid(True)
    plt.legend()
    plt.show()

    # error over lambda index for one representative path
    # pick first test path
    sample = test_subset[0]
    Fs = sample["F"]
    Ps = sample["P"]
    Psp = np.array(P_hat_batch(icnn_params, Fs))

    plt.figure(figsize=(7,4))
    plt.plot(Ps[:,0,0], label="true P11*")
    plt.plot(Psp[:,0,0], label="pred P11*")
    plt.xlabel("sample index along load path")
    plt.ylabel("P11*")
    plt.title("Task 5.2: Example load path — P11* along path")
    plt.grid(True)
    plt.legend()
    plt.show()


# ============================================================
# 5.3 PLOTS: WF model objectivity check (multi observers)
# Requires:
#   params_wf = train_WF(...)
#   P_WF_batch(params_wf, F_batch) -> (N,3,3)
# ============================================================
def random_rotations_SO3(n, seed=0):
    rng = np.random.default_rng(seed)
    Qs = []
    for _ in range(n):
        u1,u2,u3 = rng.random(3)
        q1 = np.sqrt(1-u1)*np.sin(2*np.pi*u2)
        q2 = np.sqrt(1-u1)*np.cos(2*np.pi*u2)
        q3 = np.sqrt(u1)*np.sin(2*np.pi*u3)
        q4 = np.sqrt(u1)*np.cos(2*np.pi*u3)
        x,y,z,w = q1,q2,q3,q4
        Q = np.array([
            [1-2*(y*y+z*z), 2*(x*y-z*w),   2*(x*z+y*w)],
            [2*(x*y+z*w),   1-2*(x*x+z*z), 2*(y*z-x*w)],
            [2*(x*z-y*w),   2*(y*z+x*w),   1-2*(x*x+y*y)]
        ], dtype=float)
        Qs.append(Q)
    return np.stack(Qs, axis=0)  # (n,3,3)

def left_mul(Q, A):
    # Q (3,3), A (N,3,3) -> (N,3,3)
    return np.einsum("ij,njk->nik", Q, A)

def objectivity_violation(P0, Pq, Q):
    # mean ||P(QF)-Q P(F)|| / (||P(QF)||+eps)
    QP0 = left_mul(Q, P0)
    num = np.linalg.norm((Pq - QP0).reshape(Pq.shape[0],-1), axis=1)
    den = np.linalg.norm(Pq.reshape(Pq.shape[0],-1), axis=1) + 1e-12
    return float(np.mean(num/den))

def plot_task53_objectivity(cache, params_wf, n_observers=16, max_pts=800, seed=0):
    # pick a small subset of data points
    F_all = np.concatenate([x["F"] for x in cache[:30]], axis=0)
    take = subsample_idx(F_all.shape[0], max_n=max_pts, seed=seed)
    F = F_all[take]

    P0 = np.array(P_WF_batch(params_wf, F))  # (N,3,3)

    Qs = random_rotations_SO3(n_observers, seed=seed+11)
    viols = []
    for qi,Q in enumerate(Qs):
        Fq = left_mul(Q, F)
        Pq = np.array(P_WF_batch(params_wf, Fq))
        viols.append(objectivity_violation(P0, Pq, Q))

    plt.figure(figsize=(7,4))
    plt.plot(np.arange(len(viols)), viols, "-o")
    plt.xlabel("observer rotation index")
    plt.ylabel("mean objectivity violation")
    plt.title("Task 5.3: Objectivity check for WF model")
    plt.grid(True)
    plt.show()


# ============================================================
# 5.4 PLOTS: augmentation sweep |G_obj| = 0,8,16,32
# Requires:
#   train_WF(F,P,W,epochs,lr,seed,lamW)  -> params_wf
# ============================================================
def augment_by_rotations(F, P, W, Qs):
    F_list = [F]; P_list = [P]; W_list = [W]
    for Q in Qs:
        F_list.append(left_mul(Q, F))
        P_list.append(left_mul(Q, P))
        W_list.append(W)  # W objective
    return (np.concatenate(F_list, axis=0),
            np.concatenate(P_list, axis=0),
            np.concatenate(W_list, axis=0))

def plot_task54_augmentation(cache, n_list=(0,8,16,32),
                             k_train=60, epochs=1200, lr=1e-4, seed=0,
                             n_check=16, max_pts_check=800):
    rng = np.random.default_rng(seed)
    idx = np.arange(len(cache))
    rng.shuffle(idx)
    train_ids = idx[:k_train]
    test_ids  = idx[k_train:]

    train_subset = [cache[i] for i in train_ids]
    test_subset  = [cache[i] for i in test_ids]

    Ftr = np.concatenate([x["F"] for x in train_subset], axis=0)
    Ptr = np.concatenate([x["P"] for x in train_subset], axis=0)
    Wtr = np.concatenate([x["W"] for x in train_subset], axis=0)

    Fte = np.concatenate([x["F"] for x in test_subset], axis=0)
    Pte = np.concatenate([x["P"] for x in test_subset], axis=0)

    # small check set for objectivity
    take = subsample_idx(Fte.shape[0], max_n=max_pts_check, seed=seed)
    Fchk = Fte[take]
    Pchk_true = Pte[take]

    Qcheck = random_rotations_SO3(n_check, seed=seed+222)

    mean_viol = []
    test_relmse = []

    for n_rot in n_list:
        if n_rot == 0:
            Faug, Paug, Waug = Ftr, Ptr, Wtr
        else:
            Qs = random_rotations_SO3(n_rot, seed=seed+100+n_rot)
            Faug, Paug, Waug = augment_by_rotations(Ftr, Ptr, Wtr, Qs)

        params = train_WF(Faug, Paug, Waug, epochs=epochs, lr=lr, seed=seed, lamW=1e-4)

        # test error
        Ppred = np.array(P_WF_batch(params, Fte))
        err = float(np.mean((Ppred - Pte)**2) / (np.mean(Pte**2) + 1e-12))
        test_relmse.append(err)

        # objectivity violation on check subset
        P0 = np.array(P_WF_batch(params, Fchk))
        viols = []
        for Q in Qcheck:
            Fq = left_mul(Q, Fchk)
            Pq = np.array(P_WF_batch(params, Fq))
            viols.append(objectivity_violation(P0, Pq, Q))
        mean_viol.append(float(np.mean(viols)))

        print(f"|G_obj|={n_rot:>3d} | test relMSE={err:.3e} | mean viol={mean_viol[-1]:.3e}")

    # Plot mean violation vs augmentation size
    plt.figure(figsize=(6,4))
    plt.plot(n_list, mean_viol, "-o")
    plt.xlabel("|G_obj| rotations used in augmentation")
    plt.ylabel("mean objectivity violation (check set)")
    plt.title("Task 5.4: Objectivity improves with augmentation")
    plt.grid(True)
    plt.show()

    # Plot test error vs augmentation size
    plt.figure(figsize=(6,4))
    plt.plot(n_list, test_relmse, "-o")
    plt.xlabel("|G_obj| rotations used in augmentation")
    plt.ylabel("test relative MSE(P)")
    plt.title("Task 5.4: Test error vs augmentation size")
    plt.grid(True)
    plt.show()


# ============================================================
# RUN ALL PLOTS
# ============================================================
plot_task51(cache, max_paths=60, max_pts=2000)

# 5.2 needs your invariant-model training funcs already defined
plot_task52_invariant_model(cache, k_train=50, epochs=2000, seed=0, max_test_pts=3000)

# 5.3 and 5.4 need WF model (Sec 5.3) already implemented:
params_wf = train_WF(Ftr, Ptr, Wtr, epochs=1500, lr=1e-4, seed=0, lamW=1e-4)
plot_task53_objectivity(cache, params_wf, n_observers=16, max_pts=800, seed=0)
plot_task54_augmentation(cache, n_list=(0,8,16,32), k_train=60, epochs=1200, lr=1e-4, seed=0)
```
